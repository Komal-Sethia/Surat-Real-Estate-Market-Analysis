```python
#Floor validation function
def is_valid_floor(value):
    if pd.isna(value):
        return False
    
    value = str(value).strip()
    
    pattern = r'^(Ground|Lower Basement|Upper Basement|\d+)\s+out\s+of\s+\d+$'
    
    return bool(pd.Series([value]).str.match(pattern).iloc[0])
```

```python
test_values = [
    '5 out of 13',
    'Ground out of 2',
    'Unfurnished',
    'Resale'
]

for value in test_values:
    print(value, "->", is_valid_floor(value))
```

```python
#Status validation
def is_valid_status(value):
    if pd.isna(value):
        return False
    
    value = str(value).strip()
    
    if value == 'Ready to Move':
        return True
    
    return bool(
        pd.Series([value]).str.match(
            r"^Poss\.\s+by\s+[A-Za-z]{3}\s+'\d{2}$"
        ).iloc[0]
    )
```

```python
for value in [
    'Ready to Move',
    "Poss. by Dec '25",
    'Resale',
    'Unfurnished'
]:
    print(value, "->", is_valid_status(value))
```

```python
#Price classification
df['price_is_call'] = (
    df['price']
    .astype("string")
    .str.strip()
    .eq('Call for Price')
)

df['price_is_call'].value_counts()
```

```python
#Price ko INR me numeric convert
def convert_price(value):
    if pd.isna(value):
        return np.nan
    
    value = str(value).strip()
    
    if value.lower() == 'call for price':
        return np.nan
    
    value = value.replace('₹', '').strip()
    
    if value.lower().endswith('lac'):
        number = value[:-3].strip()
        number = number.replace(',', '')
        return float(number) * 100000
    
    if value.lower().endswith('cr'):
        number = value[:-2].strip()
        number = number.replace(',', '')
        return float(number) * 10000000
    
    return np.nan
```

```python
df['price_inr'] = df['price'].apply(convert_price)
```

```python
def extract_property_type(name):
    if pd.isna(name):
        return np.nan
    
    name = str(name)
    
    for ptype in sorted(property_types, key=len, reverse=True):
        if ptype.lower() in name.lower():
            return ptype
    
    return np.nan
```
